```python
import os
import pandas as pd

def clean_crypto_csvs(folder_path="data"):
    for filename in os.listdir(folder_path):
        if filename.endswith("_hist.csv"):
            filepath = os.path.join(folder_path, filename)
            try:
                df = pd.read_csv(filepath)

                for col in ["open_price", "high", "low", "close_price", "volume", "market_cap"]:
                    if col in df.columns:
                        df[col] = df[col].replace({r'\$': '', ',': ''}, regex=True).astype(float)

                if "date" in df.columns:
                    df["date"] = pd.to_datetime(df["date"])

                df.to_csv(filepath, index=False)
                print(f"[✓] Cleaned: {filename}")

            except Exception as e:
                print(f"[!] Failed to clean {filename}: {e}")
```

```python
clean_crypto_csvs("data")
```

```python
import os
import seaborn as sns

folder_path = "data"
dfs = []
for filename in os.listdir(folder_path):
    if filename.endswith("_hist.csv"):
        coin = filename.replace("_hist.csv", "")
        df = pd.read_csv(os.path.join(folder_path, filename))
        df['coin'] = coin
        dfs.append(df)

all_crypto = pd.concat(dfs, ignore_index=True)

pivot = all_crypto.pivot_table(index='date', columns='coin', values='close_price')
corr = pivot.astype(float).corr()

plt.figure(figsize=(12, 8))
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt=".2f")
plt.title('Correlation Heatmap of Closing Prices for All Cryptos')
plt.tight_layout()
plt.show()
```
